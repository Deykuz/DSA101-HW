## Task 3

3. Report which faculty has the longest mean `days_kept`, and that mean, rounded
   to two decimals. Missing values are excluded, not treated as zero.


```python
# groupby(...).mean() already skips NaN, so missing loans are excluded rather
# than treated as zero -- which is what the task asks for.
fac_mean = loans.groupby("faculty")["days_kept"].mean()

```

```python
faculty_longest = str(fac_mean.idxmax())             # TODO: which faculty has the longest mean days_kept?
faculty_longest_mean_days = round(float(fac_mean.max()), 2) # TODO: that mean, rounded to 2 decimals


print(f"Task 3 Results -> Faculty: {faculty_longest}, Mean Days Kept: {faculty_longest_mean_days}")
```

## Task 4

```python
import matplotlib
matplotlib.use("Agg")           # no display needed for a batch run
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7.5, 4.6))
data_to_plot = [
    loans[loans['member_type'] == 'undergraduate']['days_kept'].dropna(),
    loans[loans['member_type'] == 'postgraduate']['days_kept'].dropna(),
    loans[loans['member_type'] == 'staff']['days_kept'].dropna()
]

plt.boxplot(data_to_plot, labels=['Undergraduate', 'Postgraduate', 'Staff'], patch_artist=True,
            boxprops=dict(facecolor='#a6cee3', color='#1f78b4'),
            medianprops=dict(color='darkred', linewidth=1.5))

plt.title('Distribution of Days Kept Across Member Types', fontsize=14, pad=12)
plt.xlabel('Member Type', fontsize=12)
plt.ylabel('Days Kept (excluding unlogged returns)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

output_fig_path = "figure1.png"
plt.savefig(output_fig_path, dpi=300, bbox_inches='tight')
plt.show()
print(f"Figure saved successfully to {output_fig_path}")

# TODO: draw something on `ax` that shows how days_kept is DISTRIBUTED across
# the three member types (undergraduate, postgraduate, staff) -- not just
# their means. A bar chart of three averages does not show a distribution;
# think about what does. Remember to drop the missing days_kept values rather
# than plotting them as zero.


fig.tight_layout()
fig_path = os.path.join(OUT_DIR, "figure1.png")
fig.savefig(fig_path)
plt.close(fig)
print("saved", fig_path)


```
